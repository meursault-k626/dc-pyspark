# I

## What is PySpark?

- Apache Spark = open-source, distributed computing system
  - Built for **fast processing of large-scale data**
- PySpark = the **Python interface** for Apache Spark
- Handles large datasets efficiently via **parallel computation** in Python workflows
- Good for:
  - Batch processing
  - Real-time streaming
  - Machine learning
  - Data analytics
  - SQL querying
- Used across industries — finance, healthcare, e-commerce
- Key selling points: **speed** and **scalability**

---

## When would we use PySpark?

- Best suited for datasets that are **too large to process on a single machine**
- Main use cases:
  - **Big Data Analytics** → distributed data processing
    - Uses Spark's **in-memory computation** → much faster than disk-based processing
  - **Machine Learning on large datasets**
    - Leverages Spark's **MLlib** library
    - Scalable model training + evaluation
  - **ETL and ELT Pipelines**
    - Transforms large volumes of raw data from various sources into structured formats
- Very flexible — works with lots of data sources (CSVs, Parquet, etc.)

---

## Spark Cluster

- A **cluster** = a group of computers (nodes) working together to process large datasets
- Architecture:
  - **Master node** → coordinates everything
    - Manages resources
    - Assigns tasks
  - **Worker nodes** → actually do the compute work
    - Execute the tasks given to them by the master
- This is what enables **distributed processing**

```
        [ Master Node ]
         /    |    \
   [Worker][Worker][Worker]
```

## Reading data from various formats

PySpark can read a bunch of formats — each has trade-offs.

### CSV (Comma-Separated Values)
- **Pros**: simple, plain text, works everywhere, no special tools needed
- **Cons**: **no schema enforcement** → data types aren't defined/enforced → inconsistencies creep in
- Read with:

```python
df = spark.read.csv("adult_reduced.csv", header=True, inferSchema=True)
```

### JSON (JavaScript Object Notation)
- **Pros**: great for **nested / hierarchical data**, arrays, high compatibility
- **Cons**: **storage-heavy** at scale (lots of repeated keys)
- Read with:

```python
df = spark.read.json("adults.json")
```

### Parquet
- **Columnar** storage format → optimised for **read-heavy** operations
- Great for large datasets that get queried a lot
- **Enforces schemas** → keeps data consistent
- Supports complex / nested types (like JSON)
- Read with:

```python
df = spark.read.parquet("file.parquet")
```

#### Quick comparison

| Format  | Schema enforced? | Nested data? | Best for                   |
|---------|------------------|--------------|----------------------------|
| CSV     | No               | No           | Simple, portable data      |
| JSON    | Loose            | Yes          | Hierarchical / API data    |
| Parquet | **Yes**          | Yes          | Big, read-heavy analytics  |

## Defining a schema manually

- Import the classes from `pyspark.sql.types`
- Use `StructType()` + `StructField()` to describe the structure

Example for `adult_reduced.csv`:

```python
from pyspark.sql.types import (
    StructType, StructField,
    IntegerType, StringType, DoubleType,
)

schema = StructType([
    StructField("age", IntegerType(), True),
    StructField("education.num", IntegerType(), True),
    StructField("marital.status", StringType(), True),
    StructField("occupation", StringType(), True),
    StructField("income", DoubleType(), True),
])

df = spark.read.csv("adult_reduced.csv", header=True, schema=schema)
```

- The `True` at the end of each `StructField` → whether the column is nullable
- No need for `inferSchema=True` when passing a schema explicitly

## Handling missing data

Null values are a real headache in PySpark — they can skew results or throw errors outright, so dealing with them properly is essential before any serious analysis.

There are two main approaches:

**1. Drop the nulls**

Use `.na.drop()` to remove rows containing nulls, either across the whole DataFrame or in specific columns. Quick and simple, but the warning is that if nulls are common throughout the dataset, we could end up slashing the size of our data dramatically.

For column-specific filtering, combine `.where()` with `isNotNull()`:

```python
df_cleaned = df.na.drop()
df_cleaned = df.where(col("columnName").isNotNull())
```

**2. Fill the nulls**

Use `.na.fill()` to replace nulls with a default value. This is the better choice when nulls are sparse, or when dropping rows would cost us too much data. Keeps the dataset complete and consistent.

```python
df_filled = df.na.fill({"age": 0})
```

## Column operations

PySpark gives us a tidy set of methods for manipulating columns.

- **`.withColumn()`** — add a new column, either from a calculation or derived from existing columns. Good for derived metrics and transformations.
  - Syntax: `.withColumn("new_col_name", <expression/transformation goes here>)`
  - Example: `df = df.withColumn("age_plus_5", df["age"] + 5)`

- **`.withColumnRenamed()`** — rename a column. Helpful for clarity, especially in collaborative settings where descriptive names make the DataFrame much easier to work with.
  - Syntax: `.withColumnRenamed("old_name", "new_name")`
  - Example: `df = df.withColumnRenamed("age", "years")`

- **`.drop()`** — remove unnecessary or redundant columns. Cuts down memory usage and keeps the DataFrame focused on what actually matters for the analysis.
  - Syntax: `.drop("column_name")`

Together these give the DataFrame flexibility — we can shape it to suit whatever analysis we're doing while keeping the structure clean.

**Filtering** — narrows the DataFrame down to just the rows that matter. We apply a condition to isolate subsets, e.g. data from a specific time period, a geographic region, or a particular category. Done using `.filter()`, passing in the column and the criterion.

**Grouping** — organises rows by meaningful segments (customer, product, date, etc.) so we can analyse patterns within each category. Use `.groupby()` followed by the aggregation we want on the column of interest.

Patterns that would be invisible in the raw, ungrouped data often jump out once you slice it this way.

## Joins in PySpark

Joins combine data from multiple DataFrames based on shared columns, similar to SQL. Useful for enriching datasets — e.g. merging customer details with purchase history to analyse buying patterns.

PySpark supports four join types:
- **Inner** — only matching rows from both DataFrames
- **Left** — all rows from left, matching from right
- **Right** — all rows from right, matching from left
- **Full outer** — all rows from both, with nulls where no match

Performed via the `.join()` method, specifying the second DataFrame, the join columns, and the join type:

```python
joined_df = customers.join(purchases, on="customer_id", how="inner")
```

When the join columns have **different names** across DataFrames, specify the condition explicitly (SQL-style):

```python
joined_df = customers.join(
    purchases,
    customers.customer_id == purchases.cust_id,
    how="left"
)
```

## Union operation

Union lets us combine — or "stack" — two DataFrames vertically, appending the rows of one beneath the other. Particularly useful when datasets have been split across different sources or time periods and need to be consolidated for analysis.

**Key requirement:** both DataFrames must share the **same schema** — identical number of columns, same data types, in the same order. Mismatched schemas will raise an error.

Example use case: monthly sales data stored as separate files for each month. Union combines all twelve monthly DataFrames into one DataFrame covering the whole year, removing the need to juggle multiple separate objects.

Syntax:

```python
combined_df = df1.union(df2)
```

This stacks `df2` underneath `df1`. If schemas don't align, PySpark throws an error — the operation cannot reconcile columns or types automatically.

Important to remember: union is positional, not name-based. Columns are matched by order, so column *names* alone being the same isn't enough — the order and types must match too.

## Working with Arrays and Maps

Complex data types — arrays, structs, maps — allow nested data within each row, letting a single column hold structured, hierarchical content rather than just a scalar value.

### Arrays

Arrays store lists within a column. Useful when an attribute naturally has multiple values (e.g. a list of tags, phone numbers, or skills per person).

Define values using the appropriate datatype, with `lit()` wrapping specific literal values:

```python
from pyspark.sql.functions import array, lit

df = df.withColumn(
    "skills",
    array(lit("Python"), lit("SQL"), lit("Spark"))
)
```

### Maps

Maps store dynamic key-value pairs within a column — flexible when each row might have a different set of keys (e.g. per-row metadata, variable attributes).

The `create_map()` method takes alternating key and value expressions in **key, value, key, value...** order:

```python
from pyspark.sql.functions import create_map, lit

df = df.withColumn(
    "attributes",
    create_map(
        lit("colour"), lit("red"),
        lit("size"), lit("M")
    )
)
```

Important to remember: arguments **must come in pairs**. Passing an odd number of arguments raises an `AnalysisException` — "map expects a positive even number of arguments" — because one key would be left without a value. If a value is genuinely missing, use `lit(None).cast(SomeType)` rather than omitting the argument entirely, so the pair count stays even.

## Working with Structs

Structs group related fields together within a single column — valuable for managing hierarchical data where multiple related attributes belong logically together (e.g. an address made up of street, city, and postcode).

Similar to maps, but each field is explicitly **named and typed** — the field names become part of the column's schema rather than per-row keys:

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType
from pyspark.sql.functions import struct, lit

df = df.withColumn(
    "address",
    struct(
        lit("221B Baker Street").alias("street"),
        lit("London").alias("city"),
        lit("NW1").alias("postcode")
    )
)
```

After this, the column has the schema `address: struct<street: string, city: string, postcode: string>`. Every row in the column must conform to that schema.

### Schema defined once, applies to every row

The aliases inside `struct()` aren't per-row labels — they define the field names **once**, at the column level. When adding new rows later, you don't re-alias; just supply values in the right field order (or use `Row` with named arguments):

```python
from pyspark.sql import Row

new_row = spark.createDataFrame([
    Row(address=Row(street="10 Downing St", city="London", postcode="SW1A"))
])
df = df.unionByName(new_row)
```

### Accessing nested fields

Dot notation reads a field from every row:

```python
df.select("address.city").show()          # just the city column
df.select("address.*").show()              # explode all struct fields
df.filter(df.address.postcode == "NW1")   # filter by nested field
```

Important to remember — this is the key difference between **struct** and **map**:
- **Struct** → fixed, named schema, like a mini record. Field names are identical for every row.
- **Map** → dynamic keys that can differ row to row.

Use struct when the "shape" of the nested data is known and consistent; use map when the keys themselves are data that varies.

## User Defined Functions

- What User-Defined Functions (UDFs) are in PySpark
- How to create and register them
- The two main kinds: PySpark UDFs and pandas UDFs

## UDFs for repeatable tasks

- A UDF is a custom function we create to work with data using PySpark DataFrames
- Two main kinds of UDFs to be aware of:
  - **PySpark UDFs** — suited for smaller datasets
  - **pandas UDFs** — suited for larger datasets
- The main difference between the two is the size of the dataset they are designed to handle
- There are some differences in execution, but they operate very similarly
- UDFs are reusable and repeatable because they are registered directly to the `SparkSession`
- The sizing rule may seem counterintuitive, but it comes down to how each type of UDF handles data on a row-by-row basis — worth exploring further once more PySpark experience is gained

## Defining and registering a UDF

**Step-by-step process:**

1. Create a regular Python function (e.g. `to_upper_case()`)
2. Register it as a UDF using PySpark's `.udf()` function, passing the correct data type
3. Apply it to the DataFrame

For string operations we use `StringType()`. Without registering, the UDF would not be available to all the worker nodes of the Spark Session.

```python
from pyspark.sql.functions import udf
from pyspark.sql.types import StringType

# 1. Define the Python function
def to_upper_case(s):
    return s.upper()

# 2. Register as a UDF with the appropriate return type
to_upper_case_udf = udf(to_upper_case, StringType())

# 3. Apply to the DataFrame
df.withColumn("name_upper", to_upper_case_udf(df["name"])).show()
```

Important to remember: registration to the `SparkSession` is what makes the UDF available across every worker node in the cluster.

## pandas UDF

- PySpark UDFs are useful but can introduce **performance overhead**
- This is because PySpark has to perform a series of inefficient conversions when passing data between the JVM and Python, which causes frustrating performance problems
- To mitigate this, PySpark introduced **pandas UDFs**

**Two things to notice about pandas UDFs:**

1. We have to import `pandas_udf()` directly
2. We use the `@` decorator to define the data type, e.g. `@pandas_udf("float")` — a way to take a function as an argument
3. No need to register it to the `SparkSession`

```python
from pyspark.sql.functions import pandas_udf
import pandas as pd

# Use the decorator to specify the return type
@pandas_udf("float")
def multiply_by_two(s: pd.Series) -> pd.Series:
    return s * 2

# Apply directly to a column
df.withColumn("doubled", multiply_by_two(df["value"])).show()
```

Key point: the decorator syntax removes the need for the extra registration step we saw with the PySpark UDF.

## PySpark UDFs vs. pandas UDFs

| | **PySpark UDF** | **pandas UDF** |
|---|---|---|
| Best for | Small datasets, simple transformations | Large datasets |
| Operates on | Columnar level (row-by-row in Python) | Batches of rows (vectorised via pandas) |
| Registration | Registered to `SparkSession` | Lives outside the `SparkSession` via decorator |
| Performance | Overhead from JVM↔Python conversions | Superior performance at scale |

- Registering a PySpark UDF to a `SparkSession` makes it work with all nodes of the Spark cluster
- For large datasets, pandas UDFs are preferred due to their superior performance at scale
- Choosing between them in real projects involves a cost-benefit analysis of **compute cost, development environment, data size and data type** — outside the scope of this course

## Key takeaways / cheat sheet

- **UDF** = User-Defined Function — a custom Python function made usable within PySpark DataFrames
- **PySpark UDF** → small datasets, register with `udf(func, ReturnType())`, must be registered to the `SparkSession`
- **pandas UDF** → large datasets, defined with the `@pandas_udf("type")` decorator, no registration needed
- Always specify the **return type** when defining a UDF
- pandas UDFs are vectorised, which is why they scale better — they avoid the row-by-row JVM↔Python conversion overhead
- Common PySpark types to remember: `StringType()`, `IntegerType()`, `FloatType()`, `DoubleType()`, `BooleanType()` (import from `pyspark.sql.types`)

**Quick syntax reference:**

```python

from pyspark.sql.types import StringType

# pandas UDF
from pyspark.sql.functions import pandas_udf

@pandas_udf("double")
def my_pandas_udf(s):
    return s * 1.5

df.withColumn("scaled", my_pandas_udf(df["col"]))
```

##  `Resilient Distributed Datasets (RDDs)`

## What is parallelisation in PySpark?

- One of PySpark's greatest strengths is its ability to handle large-scale data processing through **parallelisation**
- Parallelisation splits data and computations across multiple nodes in a cluster
- Operations defined in Spark are **automatically distributed** — no manual splitting required
- Tasks are assigned to **worker nodes** that process data in parallel, with results combined at the end
- This approach allows efficient processing at scale — comfortably handling gigabytes or even terabytes of data

## Understanding RDDs

- **RDDs** = Resilient Distributed Datasets — the core building blocks of Spark
- Represent distributed collections of data across a cluster
- Enable fast data access and analysis
- DataFrames offer greater user-friendliness due to simpler syntax, though they can be slower in some scenarios
- RDDs are **immutable** — once created, an RDD cannot be changed; a new RDD must be created from operations like `map()` and `filter()`
- Support **actions** like `collect()`, which retrieves the results of RDD operations

```python
# Example of immutability — operations create a new RDD rather than modifying the existing one
rdd = spark.sparkContext.parallelize([1, 2, 3, 4, 5])

# map() creates a new RDD
rdd_doubled = rdd.map(lambda x: x * 2)

# filter() creates another new RDD
rdd_even = rdd.filter(lambda x: x % 2 == 0)
```

## Creating an RDD

- Load a DataFrame from a CSV, then convert it to an RDD using the `.rdd` attribute
- Data is distributed across the cluster when the RDD is created
- Use `collect()` to retrieve and display the data

```python
# Load a DataFrame from a CSV file
df = spark.read.csv("data.csv", header=True, inferSchema=True)

# Convert the DataFrame to an RDD
rdd = df.rdd

# Retrieve all elements back to the driver
rdd.collect()
```

## Showing collect

- `collect()` shows a summary of the processes that were performed
- It is also rather verbose compared to a simple print statement — use with care, especially on large datasets, as it pulls **all** data back to the driver node

```python
# collect() returns a list of Row objects
results = rdd.collect()
for row in results:
    print(row)
```

Key point: `collect()` is fine for small RDDs or debugging, but for large datasets prefer `take(n)` to grab just a sample.

```python
# Safer alternative for large RDDs
rdd.take(5)
```

## RDDs vs DataFrames

**RDDs — low-level interface:**

- Provide **maximum flexibility** — manipulate data at a granular level
- Flexibility comes at the cost of requiring more lines of code for even moderately complex operations
- Strength: ability to **preserve data types** across operations
- Weakness: lack the schema-aware optimisations of DataFrames, making operations on structured data less efficient and harder to express
- Can scale to handle large datasets, but are not as optimised for analytics

**DataFrames — high-level abstraction:**

- Optimised for **ease of use** — encapsulate complex computations
- Achieve objectives with less code and fewer errors
- Standout feature: **SQL-like functionality** — complex transformations and analyses in just a few lines of code
- Built-in **schema awareness** — contain column names and data types, just like a structured table in SQL

| Feature | **RDD** | **DataFrame** |
|---|---|---|
| Level | Low-level | High-level abstraction |
| Flexibility | Maximum | Moderate |
| Code verbosity | More lines | Fewer lines |
| Schema awareness | No | Yes |
| Optimisation | Limited | Catalyst optimiser |
| SQL-like syntax | No | Yes |
| Data type preservation | Strong | Managed by schema |

```
┌─────────────────────────────────┐
│    DataFrame / Dataset API      │  ← high-level, declarative, schema-aware
├─────────────────────────────────┤
│            RDD API              │  ← low-level, imperative, no schema
├─────────────────────────────────┤
│   Spark Core (scheduler,        │
│   shuffle, fault tolerance)     │
└─────────────────────────────────┘
```

```python
# Same operation — RDD vs DataFrame

# RDD approach (more code)
rdd_result = rdd.map(lambda row: (row["name"], row["age"] + 1))

# DataFrame approach (concise, SQL-like)
from pyspark.sql.functions import col
df_result = df.select("name", (col("age") + 1).alias("age_plus_one"))
```

## Some useful functions and methods

- **`map()`** — applies a function to every element of an RDD. Can take a lambda or any defined/imported function
- **`collect()`** — gathers data across the cluster, leveraging PySpark's parallelisation

```python
# map() with a lambda function
rdd = spark.sparkContext.parallelize([1, 2, 3, 4])
squared = rdd.map(lambda x: x ** 2)

# map() with a named function
def add_prefix(name):
    return f"user_{name}"

names_rdd = spark.sparkContext.parallelize(["alice", "bob", "carol"])
prefixed = names_rdd.map(add_prefix)

# collect() to retrieve results
print(squared.collect())    # [1, 4, 9, 16]
print(prefixed.collect())   # ['user_alice', 'user_bob', 'user_carol']
```

## Key takeaways / cheat sheet

- **RDD** = Resilient Distributed Dataset — Spark's core low-level data structure
- **Immutable** — operations always produce a new RDD rather than modifying the original
- **Parallelisation** splits data and work across worker nodes automatically
- Convert a DataFrame → RDD with `.rdd`
- Convert back with `rdd.toDF()` when needing DataFrame features
- Use `collect()` carefully — it pulls everything to the driver; prefer `take(n)` for large RDDs
- **When to use what:**
  - **RDDs** → granular control, custom transformations, unstructured data, preserving exact types
  - **DataFrames** → structured/tabular data, SQL-style queries, better optimisation, less code

**Quick syntax reference:**

```python

# Create an RDD from a text file

```

## Intro to Spark SQL

## What is Spark SQL

- A component of Apache Spark that slots straight into its ecosystem
- Designed for processing **structured** and **semi-structured** data using SQL syntax
- Why pick it over other tools?
  - Leverages Spark's distributed computing power — handles massive datasets by spreading computation across a cluster
  - Familiar SQL syntax, so it's approachable for both analysts and engineers
  - Integrates with PySpark DataFrames, so SQL and programmatic operations can be blended in a single workflow
- In short: flexibility, speed and scalability for large, diverse datasets

## Creating temp tables

- To use SQL in PySpark, first register a DataFrame as a **temporary view**
- The DataFrame itself can come from `createDataFrame()` or from loading a flat file
- Register with `.createOrReplaceTempView("view_name")` — the view name is passed as a string
- Temporary views live only for the current Spark session — ideal for quick, session-based exploration
- Query the view with `spark.sql("...")`, passing the SQL as a string
- The result is another DataFrame, so processing can continue with PySpark afterwards

```python
from pyspark.sql import SparkSession

# Initialise Spark session
spark = SparkSession.builder.appName("Spark SQL Example").getOrCreate()

# Sample DataFrame
data = [("Alice", "HR", 30), ("Bob", "IT", 40), ("Cathy", "HR", 28)]
columns = ["Name", "Department", "Age"]
df = spark.createDataFrame(data, schema=columns)

# Register DataFrame as a temporary view
df.createOrReplaceTempView("people")

# Query using SQL
result = spark.sql("SELECT Name, Age FROM people WHERE Age > 30")
result.show()
```

## Deeper into temp views

- Typical workflow: load CSV → DataFrame → register as temp view → run SQL → optionally blend with DataFrame ops
- Load CSVs with `spark.read.csv()` (there are equivalents for JSON, Parquet, etc.)
- Registering as a temp view **safeguards the underlying data and data source** — SQL queries don't touch the original file
- Views are **session-scoped** — they disappear when the Spark session ends
- For longer-lived access:
  - **Global temporary views** — survive across sessions within the same Spark application (accessed via `global_temp.<name>`)
  - **Permanent tables** — persisted in the metastore (e.g. [Hive Table]("https://spark.apache.org/docs/latest/sql-data-sources-hive-tables.html"))

```python
# Load CSV into a DataFrame
salaries = spark.read.csv("salaries.csv", header=True, inferSchema=True)

# Session-scoped temp view
salaries.createOrReplaceTempView("salaries")

# Global temp view — persists across sessions in the same application
salaries.createGlobalTempView("salaries_global")

# Query a global temp view via the global_temp database
spark.sql("SELECT * FROM global_temp.salaries_global LIMIT 5").show()
```

## Combining SQL and DataFrame operations

- Key strength of Spark SQL: the output of `spark.sql()` is itself a DataFrame (specifically `pyspark.sql.dataframe.DataFrame`)
- This means SQL results can be fed straight into DataFrame transformations
- Example pattern: use SQL to filter, then use DataFrame API to derive new columns

```python
from pyspark.sql.functions import col

# Use SQL to filter high earners
high_earners = spark.sql("""
    SELECT job_title, salary_in_usd, experience_level
    FROM salaries
    WHERE salary_in_usd > 150000
""")

# Continue with DataFrame operations — add a Bonus column
high_earners_with_bonus = high_earners.withColumn(
    "Bonus", col("salary_in_usd") * 0.10
)

high_earners_with_bonus.show()
```

## Key takeaways / cheat sheet

| Task | Method |
|---|---|
| Create Spark session | `SparkSession.builder.appName(...).getOrCreate()` |
| Build DataFrame from Python | `spark.createDataFrame(data, schema=cols)` |
| Load CSV into DataFrame | `spark.read.csv(path, header=True, inferSchema=True)` |
| Register session-scoped view | `df.createOrReplaceTempView("name")` |
| Register cross-session view | `df.createGlobalTempView("name")` → query as `global_temp.name` |
| Run SQL query | `spark.sql("SELECT ... FROM view")` → returns a DataFrame |
| Mix SQL + DataFrame API | Chain `.withColumn()`, `.filter()`, etc. onto the SQL result |

- **Temp views are session-scoped** — gone when the session ends
- **Global temp views** live in the `global_temp` database for cross-session access
- Registering a view protects the original data/source — queries don't mutate it
- SQL query results are DataFrames → flexibility to switch between paradigms mid-pipeline
- Best of both worlds: SQL for readable, declarative querying + DataFrame API for programmatic transformations
